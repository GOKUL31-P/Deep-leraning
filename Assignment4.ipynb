```python
# =====================================================
# LSTM AND BIDIRECTIONAL LSTM FOR ELECTRICITY
# CONSUMPTION FORECASTING
# =====================================================
# 1. Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Bidirectional, Dropout
# 2. Generate Electricity Consumption Dataset
np.random.seed(42)
months = 120
time = np.arange(months)
consumption = (
100
+ 0.5 * time
+ 15 * np.sin(2 * np.pi * time / 12)
+ np.random.normal(0, 3, months)
)
df = pd.DataFrame({
'Month': pd.date_range(
start='2015-01-01',
periods=months,
freq='MS'
),
'Consumption': consumption
})
print("Dataset:")
print(df.head())
# 3. Visualize Dataset
plt.figure(figsize=(12, 5))
plt.plot(df['Month'], df['Consumption'])
plt.title("Monthly Electricity Consumption")
plt.xlabel("Month")
plt.ylabel("Consumption")

plt.grid()
plt.show()
# 4. MinMax Scaling
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(
df[['Consumption']]
)
# 5. Create Time Series Sequences
time_steps = 12
X = []
y = []
for i in range(time_steps, len(scaled_data)):
    X.append(scaled_data[i-time_steps:i, 0])
    y.append(scaled_data[i, 0])
X = np.array(X)
y = np.array(y)
# Reshape into 3D format
X = X.reshape(X.shape[0], X.shape[1], 1)
print("Input Shape:", X.shape)
print("Output Shape:", y.shape)
# 6. Train-Test Split
split = int(len(X) * 0.8)
X_train = X[:split]
X_test = X[split:]
y_train = y[:split]
y_test = y[split:]
print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
# 7. Build LSTM Model
lstm_model = Sequential([
LSTM(64, activation='tanh',
input_shape=(time_steps, 1)),
Dropout(0.2),
Dense(32, activation='relu'),
Dense(1)
])
lstm_model.compile(

optimizer='adam',
loss='mse',
metrics=['mae']
)
print("\nLSTM Model Summary")
lstm_model.summary()
# 8. Train LSTM Model
history = lstm_model.fit(
X_train,
y_train,
epochs=30,
batch_size=8,
validation_split=0.2,
verbose=1
)
# 9. Predict Using LSTM
lstm_pred = lstm_model.predict(X_test)
# Convert predictions into original scale
lstm_pred = scaler.inverse_transform(lstm_pred)
actual = scaler.inverse_transform(
y_test.reshape(-1, 1)
)
# 10. Evaluate LSTM
lstm_mse = mean_squared_error(actual, lstm_pred)
lstm_mae = mean_absolute_error(actual, lstm_pred)
print("\nLSTM RESULTS")
print("MSE:", lstm_mse)
print("MAE:", lstm_mae)
# 11. Plot LSTM Prediction
plt.figure(figsize=(12, 5))
plt.plot(actual, label='Actual Consumption')
plt.plot(lstm_pred, label='LSTM Prediction')
plt.title("LSTM Electricity Forecasting")
plt.xlabel("Months")
plt.ylabel("Electricity Consumption")
plt.legend()
plt.grid()
plt.show()
# 12. Build Bidirectional LSTM
bi_model = Sequential([
Bidirectional(

LSTM(64, activation='tanh'),
input_shape=(time_steps, 1)
),
Dropout(0.2),
Dense(32, activation='relu'),
Dense(1)
])
bi_model.compile(
optimizer='adam',
loss='mse',
metrics=['mae']
)
print("\nBidirectional LSTM Summary")
bi_model.summary()
# 13. Train Bidirectional LSTM
bi_history = bi_model.fit(
X_train,
y_train,
epochs=30,
batch_size=8,
validation_split=0.2,
verbose=1
)
# 14. Predict Using Bidirectional LSTM
bi_pred = bi_model.predict(X_test)
bi_pred = scaler.inverse_transform(bi_pred)
# 15. Evaluate Bidirectional LSTM
bi_mse = mean_squared_error(actual, bi_pred)
bi_mae = mean_absolute_error(actual, bi_pred)
print("\nBIDIRECTIONAL LSTM RESULTS")
print("MSE:", bi_mse)
print("MAE:", bi_mae)
# 16. Compare Both Models
print("\nMODEL COMPARISON")
print("LSTM MSE:", lstm_mse)
print("Bi-LSTM MSE:", bi_mse)
# 17. Plot Comparison
plt.figure(figsize=(12, 5))
plt.plot(actual, label='Actual')
plt.plot(lstm_pred, label='LSTM')

plt.plot(bi_pred, label='Bidirectional LSTM')
plt.title("LSTM vs Bidirectional LSTM")
plt.xlabel("Months")
plt.ylabel("Electricity Consumption")
plt.legend()
plt.grid()
plt.show()
# 18. Plot Training Loss
plt.figure(figsize=(10, 5))
plt.plot(history.history['loss'],
label='LSTM Training Loss')
plt.plot(bi_history.history['loss'],
label='Bi-LSTM Training Loss')
plt.title("Model Training Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.grid()
plt.show()
print("\nStructured Data Experiment Completed!")
```
